```python
from imutils  import paths
import cv2
import pandas as pd 
```

```python
folders=list(range(1,59))
```

```python
def image_transform(paths,target):
    lst=[]
    for i in paths:
        img=cv2.imread(i)
        lst.append(img.flatten())
        df=pd.DataFrame(lst)
        df['target']=target
    # print(df.shape)
    return df
```

```python
df=pd.DataFrame()
for i in folders:
    path=list(paths.list_images(rf"D:\Data Science\DataScienceProject\vehicle_sign_prediction\Images\{i}"))
    if len(path)==0:
        continue
    df_temp=image_transform(path,i)

    df=pd.concat([df,df_temp])
    print(df.shape,i)
```
